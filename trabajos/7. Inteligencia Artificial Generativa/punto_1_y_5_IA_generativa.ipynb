````python
%%writefile skills/captura-requerimientos/SKILL.md
---
name: captura-requerimientos
description: Lee documentos entregados (PDF, Word, texto, actas, correos) y extrae los requerimientos en una lista clara. Úsalo cuando el usuario suba o pegue un documento y pida entender qué hay que construir.
---

# Captura de requerimientos

## Cuándo usarlo
Cuando haya un documento de entrada (especificación, acta de reunión, correo del cliente, guía de curso) y todavía no esté claro qué hay que hacer.

## Pasos
1. Lee el documento completo antes de escribir nada.
2. Extrae cada requerimiento como una frase corta y verificable.
3. Clasifícalos en: funcionales, no funcionales (rendimiento, seguridad, costos), restricciones y supuestos.
4. Marca como **AMBIGUO** lo que se pueda interpretar de dos maneras o falte un dato.
5. Cita de dónde salió cada requerimiento (sección o página) para poder rastrearlo.
6. Cierra con una lista de preguntas para el cliente o profesor sobre lo ambiguo.

## Formato de salida
Una tabla con las columnas: ID | Requerimiento | Tipo | Fuente | Estado (claro / ambiguo).

## Reglas
- No inventes requerimientos que no estén en el documento.
- Si algo es una suposición tuya, dilo explícitamente.
- Si el documento no tiene requerimientos claros, dilo en vez de forzar una lista.
````

````python
%%writefile skills/documentar-ticket/SKILL.md
---
name: documentar-ticket
description: Convierte una idea o requerimiento en un ticket de trabajo completo con contexto, alcance, criterios de aceptación y tareas. Úsalo cuando el usuario quiera describir en detalle un trabajo por hacer antes de implementarlo.
---

# Documentar un ticket de trabajo

## Cuándo usarlo
Cuando el usuario tenga una tarea (nueva función, bug, mejora) y necesite dejarla escrita de forma que cualquier persona del equipo la pueda tomar.

## Estructura del ticket
1. **Título**: verbo + objeto, máximo 10 palabras.
2. **Contexto**: por qué existe esta tarea, qué problema resuelve.
3. **Alcance**: qué entra y, igual de importante, qué NO entra.
4. **Criterios de aceptación**: lista de condiciones verificables ("Dado... cuando... entonces...").
5. **Tareas**: pasos técnicos concretos, en orden.
6. **Dependencias y riesgos**: qué se necesita antes y qué puede salir mal.
7. **Estimación**: S / M / L con una línea de justificación.

## Reglas
- Si falta información clave, pregunta antes de escribir el ticket (máximo 3 preguntas).
- Cada criterio de aceptación debe poder responderse con sí o no.
- Guarda el resultado como `tickets/<id>-<titulo-corto>.md`.
````

````python
%%writefile skills/implementacion/SKILL.md
---
name: implementacion
description: Implementa un ticket ya documentado generando código y pruebas. Úsalo cuando haya un ticket o especificación clara y el usuario pida escribir el código.
---

# Implementación (código + pruebas)

## Cuándo usarlo
Cuando ya existe un ticket con criterios de aceptación y toca construirlo.

## Pasos
1. Lee el ticket y repite en una línea qué vas a construir. Si no coincide con lo que quería el usuario, para ahí.
2. Revisa el código existente para respetar el estilo y las librerías que ya se usan.
3. Escribe primero las pruebas, una por cada criterio de aceptación.
4. Escribe el código mínimo que haga pasar las pruebas.
5. Ejecuta las pruebas. Si fallan, corrige y vuelve a correr; no entregues código sin haberlo probado.
6. Refactoriza solo si queda código repetido o confuso.

## Reglas
- Cambios pequeños y fáciles de revisar; nada de reescribir archivos enteros sin necesidad.
- Nada de credenciales o claves en el código (usar variables de entorno).
- Funciones con nombres claros y comentarios solo donde el código no se explique solo.
- Al final, resume qué archivos se tocaron y qué pruebas se corrieron.
````

## Punto 5. ¿Se puede usar un LLM en un ESP32?

**Respuesta corta:** un LLM "de verdad" (tipo ChatGPT, Gemini o DeepSeek) no corre en un ESP32. Un modelo diminuto sí se puede, pero es más un experimento que algo útil. Para una aplicación real, lo normal es que el ESP32 le pida la respuesta a otro lado.

**Por qué no cabe.** Un ESP32 clásico tiene unos 520 KB de SRAM, 4 MB de flash (a veces 16) y un procesador de 240 MHz sin GPU. Algunas placas (ESP32-S3) traen PSRAM externa de 2 a 8 MB, y eso ayuda un poco. Un LLM necesita guardar todos sus parámetros en memoria, así que la cuenta es sencilla: parámetros × bytes por parámetro. Lo calculo abajo.

````python
# Cuánta memoria necesita cada modelo vs lo que tiene un ESP32
modelos = {
    "stories260K (juguete)": 260_000,
    "TinyStories 15M":       15_000_000,
    "GPT-2 small (124M)":    124_000_000,
    "Llama 3.2 1B":          1_000_000_000,
    "Llama 7B":              7_000_000_000,
}
bytes_por_param = {"float32": 4, "int8": 1, "int4": 0.5}

SRAM  = 0.52   # MB, ESP32 clásico
PSRAM = 8      # MB, ESP32-S3 con PSRAM (el caso bueno)

print(f"{'modelo':25s} {'fp32':>10s} {'int8':>10s} {'int4':>10s}   (MB)")
for nombre, p in modelos.items():
    fila = [p * b / 1e6 for b in bytes_por_param.values()]
    print(f"{nombre:25s} " + " ".join(f"{x:10.1f}" for x in fila))

print(f"\nSRAM ESP32 clásico: {SRAM} MB | PSRAM ESP32-S3: {PSRAM} MB\n")

for nombre, p in modelos.items():
    mejor = p * bytes_por_param["int4"] / 1e6
    if mejor <= SRAM:
        veredicto = "cabe hasta en la SRAM"
    elif mejor <= PSRAM:
        veredicto = "cabe solo en un S3 con PSRAM"
    else:
        veredicto = "no cabe ni cuantizado a 4 bits"
    print(f"{nombre:25s} -> {veredicto}")
````

Lo que sale de la tabla: solo el modelo de juguete (260 mil parámetros) cabe, y ni siquiera en un ESP32 básico sino en uno con PSRAM. Hay proyectos de la comunidad que corren un modelo así (basado en `llama2.c`, entrenado con TinyStories) en un ESP32-S3 a unos pocos tokens por segundo, pero lo único que sabe hacer es inventar cuentos cortos y sencillos. Sirve para demostrar que se puede, no para responder preguntas.

### Restricciones principales

- **Memoria:** es la barrera más grande, como se ve arriba. Además de los pesos hace falta espacio para el contexto (KV cache) y para el resto del programa.
- **Velocidad:** el cómputo es lento (sin GPU, sin aceleradores de matrices), así que la generación va a pocos tokens por segundo incluso con modelos mínimos.
- **Calidad:** un modelo tan pequeño no razona ni tiene conocimiento general; solo sirve para tareas muy acotadas.
- **Energía:** correr inferencia continua gasta batería y calienta, lo que choca con la idea de un microcontrolador de bajo consumo.
- **Almacenamiento:** hay que cargar los pesos desde flash o tarjeta SD, y eso también es lento.

### Entonces, ¿qué se hace en la práctica?

Tres opciones que sí tienen sentido:

1. **ESP32 como cliente:** manda el texto (o los datos de un sensor) por WiFi a un servidor que sí tenga el modelo, por ejemplo un PC con Ollama o una API en la nube, y recibe la respuesta. Es lo más común.
2. **Modelos pequeños de ML clásico (TinyML):** en vez de un LLM, usar redes pequeñas con TensorFlow Lite Micro para cosas como detectar una palabra clave o clasificar vibraciones de un motor. Eso sí corre bien en un ESP32.
3. **Híbrido:** el ESP32 hace el trabajo ligero local (filtrar, detectar eventos) y solo llama al LLM cuando hace falta, lo que ahorra datos y batería.

Dejo un ejemplo de la opción 1, simulando desde Python lo que haría el ESP32 al hablarle a un servidor Ollama.

````python
import requests, json

# En el ESP32 esto sería una petición HTTP POST; acá lo simulo desde Python.
# Suponiendo un servidor Ollama corriendo en la red local:
OLLAMA_URL = "http://192.168.1.50:11434/api/generate"

def preguntar_al_llm(lectura_sensor):
    prompt = (f"Un motor reporta una temperatura de {lectura_sensor} °C. "
              "En una frase, ¿es normal o hay que revisarlo?")
    cuerpo = {"model": "llama3.2:1b", "prompt": prompt, "stream": False}
    try:
        r = requests.post(OLLAMA_URL, json=cuerpo, timeout=30)
        return r.json()["response"]
    except Exception as e:
        return f"(sin conexión con el servidor: {e})"

print(preguntar_al_llm(95))
````

Y así se vería del lado del ESP32 (Arduino), solo como referencia, no se ejecuta en Colab:

```cpp
#include <WiFi.h>
#include <HTTPClient.h>

const char* ssid = "MI_WIFI";
const char* pass = "MI_CLAVE";
const char* url  = "http://192.168.1.50:11434/api/generate";

void setup() {
  Serial.begin(115200);
  WiFi.begin(ssid, pass);
  while (WiFi.status() != WL_CONNECTED) delay(500);
}

void loop() {
  float temp = 95.0;  // aquí iría la lectura real del sensor
  HTTPClient http;
  http.begin(url);
  http.addHeader("Content-Type", "application/json");
  String body = "{\"model\":\"llama3.2:1b\",\"prompt\":\"Motor a "
                + String(temp) + " C, es normal?\",\"stream\":false}";
  int code = http.POST(body);
  if (code > 0) Serial.println(http.getString());
  http.end();
  delay(60000);
}
```

### Conclusión

Un LLM completo no corre en un ESP32 por falta de memoria y de capacidad de cómputo. Se puede correr uno diminuto de juguete en las versiones con PSRAM, pero sin utilidad real. Para una aplicación seria, lo mejor es que el ESP32 se encargue de sensar y comunicar, y que el modelo viva en un servidor cercano (Ollama) o en la nube.
